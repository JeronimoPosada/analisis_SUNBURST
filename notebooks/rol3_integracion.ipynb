## Objetivo
Integrar todos los datos generados por Rol 1 y Rol 2, y mapearlos al ciclo de vida de datos según el marco DAMA DMBOK aplicado al caso SUNBURST de SolarWinds.

# Celda 2 - imports

import pandas as pd
import matplotlib.pyplot as plt

# Celda 3 - carga de CSVs

clientes = pd.read_csv("../data/clientes.csv")
instalaciones = pd.read_csv("../data/instalaciones.csv")
versiones = pd.read_csv("../data/versiones_software.csv")
eventos = pd.read_csv("../data/eventos_seguridad.csv")

print("Clientes:", clientes.shape)
print("Instalaciones:", instalaciones.shape)
print("Versiones:", versiones.shape)
print("Eventos:", eventos.shape)

# Celda 4 - vista previa 

clientes.head()
instalaciones.head()
eventos.head()

# Celda 5 - integracion con merge()

df = instalaciones.merge(clientes, on="cliente_id", how="left")
df = df.merge(versiones, on="version_id", how="left")
df = df.merge(eventos, on="instalacion_id", how="left")

print("Dataset consolidado:", df.shape)
df.head()

# Celda 6 - Markdown explicativo

## Integración de datos

Se realizó consolidación de los datasets usando las siguientes claves:

- `cliente_id`
- `version_id`
- `instalacion_id`

Esto permitió relacionar organizaciones cliente, instalaciones vulnerables, versiones de software y eventos de seguridad asociados al incidente SUNBURST.

# Celda 7 - Mapeo ciclo de vida 

mapa_ciclo = {
    "reconocimiento": "Descubrimiento",
    "acceso_normal": "Operación",
    "descarga_datos": "Uso",
    "alerta_ids": "Monitoreo",
    "movimiento_lateral": "Incidente",
    "persistencia": "Mantenimiento",
    "exfiltracion": "Compromiso",
    "conexion_c2": "Compromiso",
    "modificacion_logs": "Auditoría",
    "escalamiento_privilegios": "Incidente"
}

df["fase_ciclo_vida"] = df["tipo_evento"].map(mapa_ciclo)

df[["tipo_evento", "fase_ciclo_vida"]].head()

# Celda 8 - dashboard integrado

conteo = df["fase_ciclo_vida"].value_counts()

plt.figure(figsize=(10,6))
conteo.plot(kind="bar")
plt.title("Distribución de eventos por fase del ciclo de vida")
plt.xlabel("Fase")
plt.ylabel("Cantidad")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Celda 9 - Hallazgos claves (Markdown)

## Hallazgos clave

- Se detectaron inconsistencias de integridad referencial:
  - `cliente_id` inválidos (888, 999)
  - `version_id` inválido (99)

- Se identificaron valores faltantes:
  - `fecha_release` ausente en versión 3
  - valores vacíos en `nivel_datos_sensibles`

- Las versiones con `contiene_sunburst=True` concentraron mayor exposición al incidente.

- Los eventos críticos y anómalos se concentran principalmente en fases de:
  - Compromiso
  - Incidente
  - Monitoreo
